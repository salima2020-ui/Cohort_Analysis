```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sqlite3
from IPython.display import HTML, display

customers = pd.read_csv("olist_customers_dataset.csv")
orders = pd.read_csv("olist_orders_dataset.csv")
order_items = pd.read_csv("olist_order_items_dataset.csv")
```

```python
orders["order_purchase_timestamp"] = pd.to_datetime(orders["order_purchase_timestamp"], format = "mixed")
orders["order_month"]= orders["order_purchase_timestamp"].dt.strftime("%Y-%m-01")
conn = sqlite3.connect("sales.db")
customers.to_sql("olist_customers_dataset", conn, if_exists="replace", index=False)
orders.to_sql("olist_orders_dataset", conn, if_exists="replace", index=False)
order_items.to_sql("olist_order_items_dataset", conn, if_exists="replace", index=False)
```

```python
print("customers profiling")
display(HTML("<div style='height: 0.5cm;'></div>"))
customers.info()

print("\n")
print("orders profiling")
display(HTML("<div style='height: 0.5cm;'></div>"))
orders.info()
print("\n")
print("order_items profiling")
display(HTML("<div style='height: 0.5cm;'></div>"))
order_items.info()
```

```python
cohort_query = """
WITH first_purchase AS (
    SELECT 
        c.customer_unique_id, 
        MIN(o.order_month) AS cohort_month
    FROM olist_customers_dataset c
    JOIN olist_orders_dataset o ON c.customer_id = o.customer_id
    WHERE o.order_status = 'delivered'
    GROUP BY c.customer_unique_id
),
user_activities AS (
    SELECT 
        c.customer_unique_id,
        o.order_month AS activity_month
    FROM olist_customers_dataset c
    JOIN olist_orders_dataset o ON c.customer_id = o.customer_id
    WHERE o.order_status = 'delivered'
    GROUP BY c.customer_unique_id, o.order_month
),
cohort_size AS (
    SELECT 
        cohort_month, 
        COUNT(DISTINCT customer_unique_id) AS total_users
    FROM first_purchase
    GROUP BY cohort_month
)

SELECT 
    fp.cohort_month,
    ua.activity_month,
    (CAST(SUBSTR(ua.activity_month, 1, 4) AS INTEGER) - CAST(SUBSTR(fp.cohort_month, 1, 4) AS INTEGER)) * 12 +
    (CAST(SUBSTR(ua.activity_month, 6, 2) AS INTEGER) - CAST(SUBSTR(fp.cohort_month, 6, 2) AS INTEGER)) AS cohort_index,
    COUNT(DISTINCT fp.customer_unique_id) AS retained_users,
    cs.total_users
FROM first_purchase fp
JOIN user_activities ua ON fp.customer_unique_id = ua.customer_unique_id
JOIN cohort_size cs ON fp.cohort_month = cs.cohort_month
GROUP BY fp.cohort_month, ua.activity_month
ORDER BY fp.cohort_month, cohort_index;
"""

cohort_df = pd.read_sql_query(cohort_query, conn)
cohort_df.head(10)
```

```python
cohort_pivot = cohort_df.pivot(index='cohort_month', columns='cohort_index', values='retained_users')
cohort_sizes = cohort_pivot.iloc[:, 0]
retention_matrix = cohort_pivot.divide(cohort_sizes, axis=0)

fig, ax = plt.subplots(figsize=(14, 8))
im = ax.imshow(retention_matrix.values, cmap='Blues', vmin=0.0, vmax=0.01, aspect='auto')

ax.set_xticks(np.arange(retention_matrix.shape[1]))
ax.set_yticks(np.arange(retention_matrix.shape[0]))
ax.set_xticklabels([int(col) for col in retention_matrix.columns])
ax.set_yticklabels(retention_matrix.index)

for i in range(retention_matrix.shape[0]):
    for j in range(retention_matrix.shape[1]):
        val = retention_matrix.iloc[i, j]
        if not np.isnan(val):
            ax.text(j, i, f"{val:.1%}", ha="center", va="center", color="black" if val < 0.005 else "white", fontsize=8)

cbar = ax.figure.colorbar(im, ax=ax)
cbar.ax.set_ylabel("Retention Rate", rotation=-90, va="bottom")

plt.title('Olist E-Commerce: Cohort Retention Rate (%)', fontsize=14, pad=15)
plt.xlabel('Cohort Index (months since first purchase)', fontsize=11)
plt.ylabel('Cohort Month (first purchase month )', fontsize=11)

plt.tight_layout()
plt.show()
```

```python
# Average Retention Rate for All Months

sql = """WITH first_purchase AS (
    SELECT c.customer_unique_id, MIN(o.order_month) AS cohort_month
    FROM olist_customers_dataset c
    JOIN olist_orders_dataset o ON c.customer_id = o.customer_id
    WHERE o.order_status = 'delivered'
    GROUP BY c.customer_unique_id
),
user_activities AS (
    SELECT c.customer_unique_id, o.order_month AS activity_month
    FROM olist_customers_dataset c
    JOIN olist_orders_dataset o ON c.customer_id = o.customer_id
    WHERE o.order_status = 'delivered'
    GROUP BY c.customer_unique_id, o.order_month
),
cohort_size AS (
    SELECT cohort_month, COUNT(DISTINCT customer_unique_id) AS total_users
    FROM first_purchase
    GROUP BY cohort_month
),
monthly_rates AS (
    SELECT 
        fp.cohort_month,
        (CAST(SUBSTR(ua.activity_month, 1, 4) AS INTEGER) - CAST(SUBSTR(fp.cohort_month, 1, 4) AS INTEGER)) * 12 +
        (CAST(SUBSTR(ua.activity_month, 6, 2) AS INTEGER) - CAST(SUBSTR(fp.cohort_month, 6, 2) AS INTEGER)) AS period_number,
        COUNT(DISTINCT fp.customer_unique_id) * 100.0 / cs.total_users AS retention_pct
    FROM first_purchase fp
    JOIN user_activities ua ON fp.customer_unique_id = ua.customer_unique_id
    JOIN cohort_size cs ON fp.cohort_month = cs.cohort_month
    GROUP BY fp.cohort_month, ua.activity_month
)
SELECT 
    period_number,
    ROUND(AVG(retention_pct), 2) AS avg_retention_pct
FROM monthly_rates
GROUP BY period_number
ORDER BY period_number; """

cohort_df = pd.read_sql_query(sql, conn)

plt.figure(figsize=(12, 7))

plt.plot(
    cohort_df['period_number'], 
    cohort_df['avg_retention_pct'], 
    marker='o', 
    color='#1f77b4', 
    linewidth=2, 
    markersize=6
)

plt.title('Average Retention Curve Across All Cohorts', fontsize=13, pad=12)
plt.xlabel('Period Number (Months Elapsed Since First Purchase)', fontsize=10)
plt.ylabel('Average Retention Rate (%)', fontsize=10)

plt.xticks(cohort_df['period_number'])
plt.grid(True, linestyle='--', alpha=0.5)

for _, row in cohort_df.iterrows():
    x = row['period_number']
    y = row['avg_retention_pct']
    plt.annotate(
        f"{y:.2f}%", 
        (x, y), 
        textcoords="offset points", 
        xytext=(0, 7), 
        ha='center', 
        fontsize=8,
        weight='bold'
    )

plt.tight_layout()
plt.show()
```

```python
# Best vs Worst Cohort

sql_best_worst = """WITH first_purchase AS (
    SELECT c.customer_unique_id, MIN(o.order_month) AS cohort_month
    FROM olist_customers_dataset c
    JOIN olist_orders_dataset o ON c.customer_id = o.customer_id
    WHERE o.order_status = 'delivered'
    GROUP BY c.customer_unique_id
),
user_activities AS (
    SELECT c.customer_unique_id, o.order_month AS activity_month
    FROM olist_customers_dataset c
    JOIN olist_orders_dataset o ON c.customer_id = o.customer_id
    WHERE o.order_status = 'delivered'
    GROUP BY c.customer_unique_id, o.order_month
),
cohort_size AS (
    SELECT cohort_month, COUNT(DISTINCT customer_unique_id) AS total_users
    FROM first_purchase
    GROUP BY cohort_month
),
month_1_data AS (
    SELECT 
        fp.cohort_month,
        COUNT(DISTINCT fp.customer_unique_id) * 100.0 / cs.total_users AS retention_pct
    FROM first_purchase fp
    JOIN user_activities ua ON fp.customer_unique_id = ua.customer_unique_id
    JOIN cohort_size cs ON fp.cohort_month = cs.cohort_month
    WHERE (CAST(SUBSTR(ua.activity_month, 1, 4) AS INTEGER) - CAST(SUBSTR(fp.cohort_month, 1, 4) AS INTEGER)) * 12 +
          (CAST(SUBSTR(ua.activity_month, 6, 2) AS INTEGER) - CAST(SUBSTR(fp.cohort_month, 6, 2) AS INTEGER)) = 1
    GROUP BY fp.cohort_month
)
SELECT * FROM (
    SELECT 'Best Cohort' AS status, cohort_month, ROUND(retention_pct, 2) AS retention_pct 
    FROM month_1_data ORDER BY retention_pct DESC LIMIT 1
)
UNION ALL
SELECT * FROM (
    SELECT 'Worst Cohort' AS status, cohort_month, ROUND(retention_pct, 2) AS retention_pct 
    FROM month_1_data ORDER BY retention_pct ASC LIMIT 1
);
"""

best_worst_df = pd.read_sql_query(sql_best_worst, conn)

plt.figure(figsize=(6, 4))

plt.bar(best_worst_df['status'], best_worst_df['retention_pct'], color=['green', 'red'], width=0.4)

plt.title('Best vs Worst Cohort (Period 1 Retention Rate)')
plt.ylabel('Retention Rate (%)')

for i in range(len(best_worst_df)):
    val = best_worst_df['retention_pct'][i]
    plt.text(i, val, f"{val}%", ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

```python
revenue = """WITH first_purchase AS (
    SELECT c.customer_unique_id, MIN(o.order_month) AS cohort_month
    FROM olist_customers_dataset c
    JOIN olist_orders_dataset o ON c.customer_id = o.customer_id
    WHERE o.order_status = 'delivered'
    GROUP BY c.customer_unique_id
)
SELECT 
    fp.cohort_month,
    (CAST(SUBSTR(o.order_month, 1, 4) AS INTEGER) - CAST(SUBSTR(fp.cohort_month, 1, 4) AS INTEGER)) * 12 +
    (CAST(SUBSTR(o.order_month, 6, 2) AS INTEGER) - CAST(SUBSTR(fp.cohort_month, 6, 2) AS INTEGER)) AS period_number,
    ROUND(SUM(i.price + i.freight_value), 2) AS total_revenue
FROM first_purchase fp
JOIN olist_customers_dataset c ON fp.customer_unique_id = c.customer_unique_id
JOIN olist_orders_dataset o ON c.customer_id = o.customer_id
JOIN olist_order_items_dataset i ON o.order_id = i.order_id
WHERE o.order_status = 'delivered'
GROUP BY fp.cohort_month, period_number
ORDER BY fp.cohort_month, period_number; """

df_revenue = pd.read_sql_query(revenue, conn)

revenue_matrix = df_revenue.pivot(index='cohort_month', columns='period_number', values='total_revenue')

plt.figure(figsize=(16, 8))

im = plt.imshow(revenue_matrix, cmap='YlGn', aspect='auto')

plt.xticks(range(len(revenue_matrix.columns)), revenue_matrix.columns)
plt.yticks(range(len(revenue_matrix.index)), revenue_matrix.index)

for i in range(len(revenue_matrix.index)):
    for j in range(len(revenue_matrix.columns)):
        val = revenue_matrix.iloc[i, j]
        if not pd.isna(val) and val > 0:
            plt.text(j, i, f"${val:,.0f}", ha='center', va='center', color='black', fontsize=7)

plt.colorbar(im, label='Total Revenue ($)')
plt.title('Cohort Revenue Matrix ($)', fontsize=13, pad=12)
plt.xlabel('Period Number (Months Since First Purchase)', fontsize=10)
plt.ylabel('Cohort Month', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# Cohort Analysis Matrix

sql_pivot = """with first_purchase as (
    select 
        c.customer_unique_id, 
        min(o.order_month) as cohort_month
    from olist_customers_dataset c
    join olist_orders_dataset o on c.customer_id = o.customer_id
    where o.order_status = 'delivered'
    group by c.customer_unique_id
),
user_activities as (
    select 
        c.customer_unique_id, 
        o.order_month as activity_month
    from olist_customers_dataset c
    join olist_orders_dataset o on c.customer_id = o.customer_id
    where o.order_status = 'delivered'
    group by c.customer_unique_id, o.order_month
),
periods as (
    select 
        fp.cohort_month,
        fp.customer_unique_id,
        (cast(substr(ua.activity_month, 1, 4) as integer) - cast(substr(fp.cohort_month, 1, 4) as integer)) * 12 +
        (cast(substr(ua.activity_month, 6, 2) as integer) - cast(substr(fp.cohort_month, 6, 2) as integer)) as period_number
    from first_purchase fp
    join user_activities ua on fp.customer_unique_id = ua.customer_unique_id
)
select 
    cohort_month,
    count(distinct case when period_number = 0 then customer_unique_id end) as m_0,
    count(distinct case when period_number = 1 then customer_unique_id end) as m_1,
    count(distinct case when period_number = 2 then customer_unique_id end) as m_2,
    count(distinct case when period_number = 3 then customer_unique_id end) as m_3,
    count(distinct case when period_number = 4 then customer_unique_id end) as m_4,
    count(distinct case when period_number = 5 then customer_unique_id end) as m_5,
    count(distinct case when period_number = 6 then customer_unique_id end) as m_6
from periods
group by cohort_month
order by cohort_month;"""

df_pivot = pd.read_sql_query(sql_pivot, conn)

matrix_data = df_pivot.set_index('cohort_month')

plt.figure(figsize=(12, 8))

im = plt.imshow(matrix_data, cmap='YlGnBu', aspect='auto')

plt.xticks(range(len(matrix_data.columns)), matrix_data.columns)
plt.yticks(range(len(matrix_data.index)), matrix_data.index)

for i in range(len(matrix_data.index)):
    for j in range(len(matrix_data.columns)):
        val = matrix_data.iloc[i, j]
        if val > 0:
            color = 'white' if val > (matrix_data['m_0'].max() / 2) else 'black'
            plt.text(j, i, str(val), ha='center', va='center', color=color, fontsize=8)

plt.colorbar(im, label='Active Customer Count')
plt.title('Pivoted Cohort Matrix (Customer Counts)', fontsize=13, pad=12)
plt.xlabel('Periods (m_0 to m_6)', fontsize=10)
plt.ylabel('Cohort Month', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# Retention for States

sql_state_retention = """with first_purchase as (
    select 
        c.customer_unique_id, 
        c.customer_state,
        min(o.order_month) as cohort_month
    from olist_customers_dataset c
    join olist_orders_dataset o on c.customer_id = o.customer_id
    where o.order_status = 'delivered'
    group by c.customer_unique_id, c.customer_state
),
user_activities as (
    select 
        c.customer_unique_id, 
        o.order_month as activity_month
    from olist_customers_dataset c
    join olist_orders_dataset o on c.customer_id = o.customer_id
    where o.order_status = 'delivered'
    group by c.customer_unique_id, o.order_month
)
select 
    fp.customer_state,
    fp.cohort_month,
    (cast(substr(ua.activity_month, 1, 4) as integer) - cast(substr(fp.cohort_month, 1, 4) as integer)) * 12 +
    (cast(substr(ua.activity_month, 6, 2) as integer) - cast(substr(fp.cohort_month, 6, 2) as integer)) as period_number,
    count(distinct fp.customer_unique_id) as retained_users
from first_purchase fp
join user_activities ua on fp.customer_unique_id = ua.customer_unique_id
group by fp.customer_state, fp.cohort_month, period_number
order by fp.customer_state, fp.cohort_month, period_number;"""

df_state = pd.read_sql_query(sql_state_retention, conn)

top_5_states = df_state[df_state['period_number'] == 0].groupby('customer_state')['retained_users'].sum().nlargest(5).index

df_p1 = df_state[(df_state['period_number'] == 1) & (df_state['customer_state'].isin(top_5_states))]

plt.figure(figsize=(14, 6))

for state in top_5_states:
    state_data = df_p1[df_p1['customer_state'] == state]
    plt.plot(state_data['cohort_month'], state_data['retained_users'], marker='o', label=state)

plt.title('Top 5 States - Retained Users in Period 1 (Month 1)', fontsize=13, pad=12)
plt.xlabel('Cohort Month', fontsize=10)
plt.ylabel('Retained User Count (Period 1)', fontsize=10)
plt.xticks(rotation=45)
plt.legend(title='State')
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Cumulative Items Purchased per User

sql_items_per_user = """with first_purchase as (
    select 
        c.customer_unique_id,
        min(o.order_month) as cohort_month
    from olist_customers_dataset c
    join olist_orders_dataset o on c.customer_id = o.customer_id
    where o.order_status = 'delivered'
    group by c.customer_unique_id
),
cohort_size as (
    select 
        cohort_month, 
        count(distinct customer_unique_id) as total_users 
    from first_purchase
    group by cohort_month
), 
monthly_items as (
    select 
        fp.cohort_month,
        (cast(substr(o.order_month, 1, 4) as integer) - cast(substr(fp.cohort_month, 1, 4) as integer)) * 12 +
        (cast(substr(o.order_month, 6, 2) as integer) - cast(substr(fp.cohort_month, 6, 2) as integer)) as period_number,
        count(i.order_item_id) as total_items
    from first_purchase fp
    join olist_customers_dataset c on fp.customer_unique_id = c.customer_unique_id
    join olist_orders_dataset o on c.customer_id = o.customer_id
    join olist_order_items_dataset i on o.order_id = i.order_id
    where o.order_status = 'delivered'
    group by fp.cohort_month, period_number
) select 
    m.cohort_month,
    m.period_number,
    round(sum(m.total_items) over (partition by m.cohort_month order by m.period_number) * 1.0 / cs.total_users, 2) as cumulative_items_per_user
from monthly_items m
join cohort_size cs on m.cohort_month = cs.cohort_month
where m.period_number <= 6
order by m.cohort_month, m.period_number;"""

df_items = pd.read_sql_query(sql_items_per_user, conn)

matrix_items = df_items.pivot(index='cohort_month', columns='period_number', values='cumulative_items_per_user')

plt.figure(figsize=(12, 8))

im = plt.imshow(matrix_items, cmap='Blues', aspect='auto')

plt.xticks(range(len(matrix_items.columns)), matrix_items.columns)
plt.yticks(range(len(matrix_items.index)), matrix_items.index)

for i in range(len(matrix_items.index)):
    for j in range(len(matrix_items.columns)):
        val = matrix_items.iloc[i, j]
        if not pd.isna(val):
            plt.text(j, i, f"{val:.2f}", ha='center', va='center', color='black', fontsize=8)

plt.colorbar(im, label='Cumulative Items / User')
plt.title('Cumulative Items Purchased per User by Cohort (Periods 0-6)', fontsize=13, pad=12)
plt.xlabel('Period Number (Months Since First Purchase)', fontsize=10)
plt.ylabel('Cohort Month', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
# Common Freight Value for Cohort_Month

sql_freight = """
with first_purchase as (
    select 
        c.customer_unique_id,
        min(o.order_month) as cohort_month
    from olist_customers_dataset c
    join olist_orders_dataset o on c.customer_id = o.customer_id
    where o.order_status = 'delivered'
    group by c.customer_unique_id
), 
cohort_size as (
    select 
        cohort_month, 
        count(distinct customer_unique_id) as total_users 
    from first_purchase
    group by cohort_month
), 
monthly_freight as (
    select 
        fp.cohort_month,
        (cast(substr(o.order_month, 1, 4) as integer) - cast(substr(fp.cohort_month, 1, 4) as integer)) * 12 +
        (cast(substr(o.order_month, 6, 2) as integer) - cast(substr(fp.cohort_month, 6, 2) as integer)) as period_number,
        sum(i.freight_value) as total_freight
    from first_purchase fp
    join olist_customers_dataset c on fp.customer_unique_id = c.customer_unique_id
    join olist_orders_dataset o on c.customer_id = o.customer_id
    join olist_order_items_dataset i on o.order_id = i.order_id
    where o.order_status = 'delivered'
    group by fp.cohort_month, period_number
)
select 
    mf.cohort_month,
    mf.period_number,
    round(
        sum(mf.total_freight) over (
            partition by mf.cohort_month 
            order by mf.period_number
        ) * 1.0 / cs.total_users, 
    2) as cumulative_freight_per_user
from monthly_freight mf
join cohort_size cs on mf.cohort_month = cs.cohort_month
where mf.period_number <= 6
order by mf.cohort_month, mf.period_number;
"""

df_freight = pd.read_sql_query(sql_freight, conn)

freight_matrix = df_freight.pivot(index='cohort_month', columns='period_number', values='cumulative_freight_per_user')

plt.figure(figsize=(12, 8))

im = plt.imshow(freight_matrix, cmap='YlOrRd', aspect='auto')

plt.xticks(range(len(freight_matrix.columns)), freight_matrix.columns)
plt.yticks(range(len(freight_matrix.index)), freight_matrix.index)

for i in range(len(freight_matrix.index)):
    for j in range(len(freight_matrix.columns)):
        val = freight_matrix.iloc[i, j]
        if not pd.isna(val):
            plt.text(j, i, f"${val:.2f}", ha='center', va='center', color='black', fontsize=8)

plt.colorbar(im, label='Cumulative Freight Spend / User ($)')
plt.title('Cumulative Freight Spend per User by Cohort (Periods 0-6)', fontsize=13, pad=12)
plt.xlabel('Period Number (Months Since First Purchase)', fontsize=10)
plt.ylabel('Cohort Month', fontsize=10)

plt.tight_layout()
plt.show()
```
